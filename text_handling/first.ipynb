```python
from pathlib import Path

DATA_DIRECTORY=Path("data")
LOG_FILE=DATA_DIRECTORY/"order_activity.txt"

DATA_DIRECTORY.mkdir(parents=True,exist_ok=True)

initial_activities=(
    "ORD101 | Order Placed\n"
    "ORD101 | Payment completed\n"
)

LOG_FILE.write_text(
    initial_activities,
    encoding="utf-8"
)
```

```python
def record_activity(order_id,activity):
    with LOG_FILE.open("a",encoding="utf-8") as file:
        file.write(f"{order_id} | {activity}\n")
        
record_activity("O101","Order Shipped")
record_activity("O101","Order placed")
```

```python
def find_order_activities(order_id):
    if not LOG_FILE.exists():
        print("Order activity log does not exist.")
        return
    found=False
    
    with LOG_FILE.open("r",encoding="utf-8") as file:
        for activity in file:
            if activity.startswith(f"{order_id} |"):
                print(activity.strip())
                found=True
        if not found:
            print(f"Order activity of {order_id} not found")

find_order_activities("ORD101")
    
```

```python
from pathlib import Path

LOG_FILE=Path("data")/"order_activity.txt"

try:
    with LOG_FILE.open(
        "r",
        encoding="utf-8"
    ) as file:
        for activity in file:
            print(activity.strip())
except FileNotFoundError:
    print("Order activity file does not exist")
```

```python
def delete_order_activities(order_id):
    if not LOG_FILE.exists():
        print("Order activity file does not exist.")
    with LOG_FILE.open(
        mode="r",
        encoding="utf-8"
    ) as file:
        activities=file.readlines()
        
    remaining_activities=[activity for activity in activities if not activity.startswith(f"{order_id} |")]
    
    with LOG_FILE.open(
        mode="w",
        encoding="utf-8"
    ) as file:
        file.writelines(remaining_activities)
    
    print(f"Activities for {order_id} deleted.")

delete_order_activities("O101")
```

```python
DATA_DIRECTORY=Path("data")

REVIEW_FILE=DATA_DIRECTORY/"product_review.txt"

if not DATA_DIRECTORY.exists():
    DATA_DIRECTORY.mkdir(parents=True,exist_ok=True)
elif not REVIEW_FILE.exists():
    REVIEW_FILE.touch()
```

```python
def add_review(product_id,customer_name,rating,comment):
    if (not rating.isnumeric()) or (int(rating)<1 or int(rating)>5):
        print("Invalid Rating")
        return
    
    try:
        with REVIEW_FILE.open("a",encoding="utf-8") as file:
            file.write(f"{product_id} | {customer_name} | {rating}/5 | {comment}\n")
            print(f"Review of Product ID:{product_id} is added Successfully")

    except FileNotFoundError:
        print("REVIEW FILE DOES NOT EXIST")
```

```python
add_review("P101","Ankit","5","Excellent wireless Mouse")
```

```python
add_review("P102","Meera","4","Good keyboard for the price")
add_review("P101","Ravi","3","Product is good but delievery was late")
```

```python
def find_product_reviews(product_id):
    found=False
    try:
        with REVIEW_FILE.open("r",encoding="utf-8") as file:
            for review in file:
                if review.startswith(f"{product_id} |"):
                    print(review)
                    found=True
            if not found:
                print(f"No Reviews is available for Product ID:{product_id}")
    except FileNotFoundError:
        print("Review file does not exist")
                    
```
